## Ejercicio 2: Agente de soporte técnico

### Ficha PEAS

| Elemento | Descripción |
|---|---|
| **Percepción (S)** | Tiempo de espera del ticket en minutos, nivel de urgencia del problema y condición del cliente (premium o no premium). |
| **Acciones (A)** | Asignar prioridad baja, media o alta al ticket de soporte. |
| **Entorno (E)** | Sistema de atención de soporte técnico, clientes, tickets registrados y personal encargado de resolver las incidencias. |
| **Objetivo** | Priorizar los tickets de soporte considerando la urgencia, el tiempo de espera y la condición premium del cliente. |
| **Medida de desempeño (P)** | Priorizar adecuadamente los tickets urgentes y reducir tiempos de espera, considerando también la atención preferencial de clientes premium. |

### Justificación de las reglas

Para esta simulación se consideran tres factores: nivel de urgencia, tiempo de espera y condición premium del cliente.

- **Prioridad alta:** urgencia alta, o urgencia media con espera mayor o igual a 30 minutos, o cliente premium con espera mayor o igual a 30 minutos.
- **Prioridad media:** urgencia media con espera menor a 30 minutos, urgencia baja con espera mayor o igual a 30 minutos, o cliente premium que no haya alcanzado una condición de prioridad alta.
- **Prioridad baja:** urgencia baja, espera menor a 30 minutos y cliente no premium.

Estas reglas buscan que la urgencia tenga mayor influencia en la decisión, mientras que el tiempo de espera permite aumentar la prioridad de tickets que llevan más tiempo pendientes. La condición premium también incrementa la prioridad de atención. El valor de 30 minutos se establece únicamente como umbral para esta simulación.

### Código

```python
def agente_soporte(tiempo_espera_minutos, nivel_urgencia, cliente_premium):
    # Validar los datos de entrada
    if tiempo_espera_minutos < 0:
        return "error", "el tiempo de espera no puede ser negativo"

    if nivel_urgencia not in ["baja", "media", "alta"]:
        return "error", "el nivel de urgencia debe ser baja, media o alta"

    # Prioridad alta por urgencia
    if nivel_urgencia == "alta":
        return "prioridad alta", "el ticket presenta un nivel de urgencia alto"

    # Prioridad alta por combinación de espera y otras condiciones
    if tiempo_espera_minutos >= 30:
        if nivel_urgencia == "media":
            return "prioridad alta", "urgencia media con 30 minutos o más de espera"

        if cliente_premium:
            return "prioridad alta", "cliente premium con 30 minutos o más de espera"

    # Prioridad media
    if nivel_urgencia == "media":
        return "prioridad media", "urgencia media con menos de 30 minutos de espera"

    if tiempo_espera_minutos >= 30:
        return "prioridad media", "urgencia baja con 30 minutos o más de espera"

    if cliente_premium:
        return "prioridad media", "cliente premium con tiempo de espera menor a 30 minutos"

    # Prioridad baja
    return "prioridad baja", "urgencia baja, espera menor a 30 minutos y cliente no premium"
```
